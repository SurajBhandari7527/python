<h3>Adding titles and labels to the graph</h3>

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title("Rohit Sharma Vs Virat Kohli performance")
plt.xlabel("Season/Year")
plt.ylabel("Runs scored")
```

<h3>If we want to change the color of the graphs for each player(whatever data) we can use hex color code as our wish</h3>

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#000000') #extra parameter to fix color
plt.plot(batsman['index'],batsman['RG Sharma'],color='red')
plt.title("Rohit Sharma Vs Virat Kohli performance")
plt.xlabel("Season/Year")
plt.ylabel("Runs scored")
```

<h3>There are more parameters, that is 'linestyle'</h3>

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#000000',linestyle='dashed') #dotted is also there
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='dashdot',linewidth=5)
plt.title("Rohit Sharma Vs Virat Kohli performance")
plt.xlabel("Season/Year")
plt.ylabel("Runs scored")
```

<h3> There is one more parameter called marker</h3>
<h4>->It marks the edges/sharps of the graphs and its size can also be controlled</h4>


```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#000000',linestyle='dashed') #dotted is also there
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='dashdot',linewidth=2,marker='p',markersize=20)
plt.title("Rohit Sharma Vs Virat Kohli performance")
plt.xlabel("Season/Year")
plt.ylabel("Runs scored")

#markers- D for diamond shape , +,-,O,p for pentagon(whatever we want)
```

<h4>To define which color graph represents what data, we can use legend to denote the graph lines using different colors</h4>

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#000000',label='Virat') #extra parameter to fix color
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',label='Rohit')
plt.title("Rohit Sharma Vs Virat Kohli performance")
plt.xlabel("Season/Year")
plt.ylabel("Runs scored")
plt.legend(loc='upper left')
#Note: Need to include label="Name_to_define_color" in the argument and use .legend() method
```

<h4> We can limit the range of the values of x-axis and y-axis by using the method xlim and ylim </h4>

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#000000',label='Virat') #extra parameter to fix color
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',label='Rohit')
plt.title("Rohit Sharma Vs Virat Kohli performance")
plt.xlabel("Season/Year")
plt.ylabel("Runs scored")
plt.legend(loc='upper left')
plt.xlim(2008,2014)
plt.ylim(0,600)
#this is used when there is very high value in the data (outlier value) due to which the
#the small values graph is lost (to see that small values graph we can limit ranges)
```

<h4>Include grids in the graph</h4>

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#000000',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',label='Rohit')
plt.title("Rohit Sharma Vs Virat Kohli performance")
plt.xlabel("Season/Year")
plt.ylabel("Runs scored")
plt.legend(loc='upper left')
#Just use 
plt.grid()
```

<h4>Add a size dimension to your scatter plot</h4>
Basically, it adjusts the size of the dots according the parameter selected, if parameter has high value then big dot else small dot

```python
tips = sns.load_dataset('tips')
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
plt.title('Total bill vs tip given')
plt.xlabel('Total bill')
plt.ylabel('Tips given')
plt.show()
```

<h5>Note: The more the size of dots, it denotes that customer came with large no. of friends</h5>

<h3>But But, This scatter plot can also be done by using plt.plot if 'o' argument is passed</h3>

```python
tips = sns.load_dataset('tips')
plt.plot(tips['total_bill'],tips['tip'],'o')
plt.title('Total bill vs tip given')
plt.xlabel('Total bill')
plt.ylabel('Tips given')
plt.show()
```

<h4>But the drawbacks is:</h4>
1. Can't use size of dots feature

<h4>Vertical Bar chart </h4>

```python
#Vertical Bar chart
students=[20,30,10,60,40,50]
colors=['Red','Blue','Yellow','Pink','Green','White']
plt.bar(colors,students)
plt.show()
```

<h4>Horizontal Bar chart</h4>

```python
#Horizontal Bar chart (Generally used when large no. of categories)
students=[20,30,10,60,40,50]
colors=['Red','Blue','Yellow','Pink','Green','White']
plt.barh(colors,students)
plt.show()
```

```python
#Real data example
df=pd.read_csv(r"E:\python\Datascience\datasets\batsman_season_record.csv")
plt.bar(df['batsman'],df['2017'])
plt.show()
```

<h4>Jugaad method to print all the graphs (2015,2016,2017) at once</h4>

```python
plt.bar(np.arange(df.shape[0])-0.25,df['2015'],width=0.25)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.25)
plt.bar(np.arange(df.shape[0])+0.25,df['2017'],width=0.25)
#actually we are taking different points on x-axis to place the bars at diff locations
#i.e. -0.25 shifted for the first,0 shifted for second, +0.25 shifted for third
```

```python
plt.bar(np.arange(df.shape[0])-0.25,df['2015'],width=0.25,label='2015')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.25,label='2016')
plt.bar(np.arange(df.shape[0])+0.25,df['2017'],width=0.25,label='2017')
#actually we are taking different points on x-axis to place the bars at diff locations
#i.e. -0.25 shifted for the first,0 shifted for second, +0.25 shifted for third
#now, we will change the name of the x-axis data
plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.legend()
plt.show()
```

<h4>If the name os the x-axis datas are very long then It overlaps</h4>

```python
students=[20,30,10,60,40,50]
colors=['Red Red Red Red Red','Blue Blue Blue Blue','Yellow Yellow Yellow Yellow','Pink Pink Pink','Green Green Green','White White White']
plt.bar(colors,students)
plt.show()
```

<h4> To solve this problem, we can use rotation for x-axis data name</h4>

```python
students=[20,30,10,60,40,50]
colors=['Red Red Red Red Red','Blue Blue Blue Blue','Yellow Yellow Yellow Yellow','Pink Pink Pink','Green Green Green','White White White']
plt.bar(colors,students)
plt.xticks(rotation='vertical')
plt.show()
```

We can see the data names of the x-axis is printed vertically

<h3>What if we want to show all the datas in one strip at diff levels</h3>

```python
df=pd.read_csv(r"E:\python\Datascience\datasets\batsman_season_record.csv")
plt.bar(df['batsman'],df['2015'],label='2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'],label='2016')
plt.bar(df['batsman'],df['2017'],bottom=(df['2015'] + df['2016']),label='2017')
plt.legend()
plt.show()
```

<h1>Advanced Matplotlib</h1>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

<h3>
Colored scattered plot ( Groupwise)
</h3>

```python
iris=pd.read_csv(r"E:\python\Datascience\datasets\iris.csv")
pd.set_option('future.no_silent_downcasting', True) #without this we can get error because when we are replacing a column of dataframe into integer type 
#it is downcasting the object column into integer automatically which can be depreciated in future. To suppress the warning we need to use this.
iris['Species']= iris['Species'].replace({'Iris-setosa':0,'Iris-virginica':1,'Iris-versicolor':2})
iris
```

```python
plt.style.use('fivethirtyeight')
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species']) #c means color(for each group) acc to whether it belongs to species 0,1,2. same have same color
```

```python
plt.style.use('fivethirtyeight')
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],alpha=0.5) #alpha is the opacity of the plot points
```

<h4> We can use different themes of matplotlib using cmap</h4>

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='winter')
plt.colorbar() #this function specifies which color corresponds to which values
```

<h4> How to edit/set the size of the plot graph ? </h4>

```python
plt.figure(figsize=(16,9))
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='winter')
plt.colorbar()
```
